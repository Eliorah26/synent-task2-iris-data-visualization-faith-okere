```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
```

```python
column_names = [
    'sepal_length',
    'sepal_width',
    'petal_length',
    'petal_width',
    'species'
]

iris = pd.read_csv('iris.data', header=None, names=column_names)

iris.head()
```

```python
average_petal = iris.groupby('species')['petal_length'].mean()

average_petal.plot(kind='bar')

plt.title('Average Petal Length by Iris Species')
plt.xlabel('Iris Species')
plt.ylabel('Average Petal Length (cm)')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(iris['sepal_length'], bins=10, edgecolor='black')

plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

plt.title('Sepal Length vs Petal Length by Species')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')

plt.show()
```

```python
iris.groupby('species')[[
    'sepal_length',
    'sepal_width',
    'petal_length',
    'petal_width'
]].mean()
```

```python
species_comparison = iris.groupby('species')[[
    'sepal_length',
    'sepal_width',
    'petal_length',
    'petal_width'
]].mean()

species_comparison.to_csv('iris_species_comparison.csv')

files.download('iris_species_comparison.csv')
```
